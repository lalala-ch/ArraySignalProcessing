
# <span style="color:rgb(213,80,0)">Conventional Beamforming</span>

$$\begin{array}{l} P(\theta )={w^H }d(\theta ),\\ whered(\theta )=[1,e^{j\pi sin\theta } ,e^{j2\pi sin\theta } ,e^{j3\pi sin\theta } ,...,e^{j(M-1)\pi sin\theta } ] \end{array}$$

When $\theta =0$ holds, $d(\theta )=[1,1,...,1]$ ,


we set the beamformer weights ${w^H }$ to the conjugate of the steering vector to obtain a conventional beamformer.


Assume a 10\-element uniform linear array with a wavelength of 3 cm and half\-wavelength element spacing.


```matlab
lambda = 3e-2;
f=wavelen2freq(lambda)
```

```matlab
num_sensors = 10;
d = 0:lambda/2:(num_sensors-1)*lambda/2;
```


Create the steering vector


```matlab
theta = -90:1:90;
A = exp(-1i*2*pi*d.'*sin(deg2rad(theta))/lambda);
```


Determine the beamformer's complex weight vector from the signal direction. For a conventional beamformer, the weights are the steering vector.


```matlab
direction = 0;
w = exp(-1i*2*pi*d.'*sin(deg2rad(direction))/lambda);
P = w' * A;

P_abs = abs(P);
P_abs = P_abs / max(P_abs);

BP = 20*log10(max(P_abs, 1e-10));

figure;
plot(theta, BP, 'LineWidth', 1.5);
grid on;
xlim([-90 90]);
ylim([-80 0]);

xlabel('\theta (degree)');
ylabel('Normalized magnitude response (dB)');
title('CBF Narrowband Beamformer');
```


If the element spacing is increased from half a wavelength to one wavelength, spatial aliasing becomes clearly visible, with an additional main lobe in the endfire direction.


```matlab
d = 0:lambda:(num_sensors-1)*lambda;
theta = -90:1:90;
A = exp(-1i*2*pi*d.'*sin(deg2rad(theta))/lambda);
direction = 0;
w = exp(-1i*2*pi*d.'*sin(deg2rad(direction))/lambda);
P = w' * A;

P_abs = abs(P);
P_abs = P_abs / max(P_abs);

BP = 20*log10(max(P_abs, 1e-10));

figure;
plot(theta, BP, 'LineWidth', 1.5);
grid on;
xlim([-90 90]);
ylim([-80 0]);

xlabel('\theta (degree)');
ylabel('Normalized magnitude response (dB)');
title('CBF Narrowband Beamformer');
```


Increasing the number of elements makes the main lobe noticeably narrower.


```matlab
num_sensors = 20;
d = 0:lambda/2:(num_sensors-1)*lambda/2;
theta = -90:1:90;
A = exp(-1i*2*pi*d.'*sin(deg2rad(theta))/lambda);
direction = 0;
w = exp(-1i*2*pi*d.'*sin(deg2rad(direction))/lambda);
P = w' * A;

P_abs = abs(P);
P_abs = P_abs / max(P_abs);

BP = 20*log10(max(P_abs, 1e-10));

figure;
plot(theta, BP, 'LineWidth', 1.5);
grid on;
xlim([-90 90]);
ylim([-80 0]);

xlabel('\theta (degree)');
ylabel('Normalized magnitude response (dB)');
title('CBF Narrowband Beamformer');
```
